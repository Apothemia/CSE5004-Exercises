```python
import networkx as nx
```

```python
G = nx.les_miserables_graph()
assert isinstance(G, nx.Graph)

nx.draw(G, node_size=40, with_labels=True)
```

```python
# Density
nx.density(G)
```

```python
# Diameter
nx.diameter(G)
```

```python
# Degree
sorted(nx.degree(G), key=lambda n: n[1], reverse=True)[:10]
```

```python
# Betweenness
sorted(nx.betweenness_centrality(G).items(), key=lambda n: n[1], reverse=True)[:10]
```

```python
# Closeness
sorted(nx.closeness_centrality(G).items(), key=lambda n: n[1], reverse=True)[:10]
```

```python
# Eigenvector
sorted(nx.eigenvector_centrality(G).items(), key=lambda n: n[1], reverse=True)[:10]
```

```python
# Eccentricity
sorted(nx.eccentricity(G).items(), key=lambda n: n[1], reverse=True)[:10]
```
